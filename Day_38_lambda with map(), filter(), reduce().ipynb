```python
num = [1,2,4,5,7,10]
num
```

```python
# map(function, iterable objective)

num = list(map(lambda x:x**2,num))
num
```

```python
# filter(function, iterable objective)

num = list(filter(lambda num: num%2==1,num))
num
```

```python
#reduce

from functools import reduce
reduce(lambda a,b: a+b, num )
```
