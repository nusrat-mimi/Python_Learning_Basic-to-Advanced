Question: Create a function named remove_duplicates. It will take a list (input_list) as its parameter. It then iterates through the element of the input list and append each unique element to a new list (unique_list). The new list is then returned.

```python
def remove_duplicates(input_list):
    unique_list = []

    for item in input_list:
        if item not in unique_list:
            unique_list.append(item)
    return unique_list       
```

```python
input_list = [1,2,3,4,2,3,2,7,8]
remove_duplicates(input_list)

print(f'Real List{input_list}')
print(f'Unique List {remove_duplicates(input_list)}')
```
