# Functions in Python

# *1. Understanding functions*
A Function is a block of organized, reusable code that performs a specific code.
Functions help in making your code modular, more readable and easier to maintain.

# *2. Function Definition*
To define a function use key word 'def' followed by function name, parenthesis and colon. e.g:

```python
def dept():
    print('data science!')
```

# *3. Function call*
to execute a function, call it with its name and parenthesis

```python
dept()
```

# *4. Parameters*
Functions can accept parameters(input). Parameters are variables that are used as placeholders for the actual values when the function is called 

```python
def dept_with_names(names):
    print(f'My Major is {names}')
```

```python
# You can call this function with a specific value
```

```python
dept_with_names('Machine learning and Ai')
```

# *5. Return Statement*
Functions can return a value using a 'return' statement. This is useful when you want to use the results of a function elsewhere in your code

```python
def name():
    print('mimi')
    return 1
```

```python
name()
```

```python
def name():
    return print('mimi')
```

```python
name()
```

```python
def name():
    return dept()
```

```python
name()
```

# Question 2
write a function that will take two parameters, a and b, and returns their sum. The values of a and b will not fixed within the function; they are determined by the arguments passed when the function is called. That will make the function more flexible because it can add any two numbers, not just the two values in the first function.

```python
def add_num(a, b):
    return a+b
    
```

```python
add_num(10, 20)
```

```python
add_num(50,40)
```
