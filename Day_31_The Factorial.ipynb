# The Factorial

```python
def fact(x):
    result = 1
    if x == 1 or x == 0:
        print(f'The {x}! is = 1')

    else:
        for i in range(x):
            result = result * (i+1)

        return result
```

```python
fact(2)
```

```python
fact(3)
```

```python
fact(5)
```

```python
fact(0)
```
