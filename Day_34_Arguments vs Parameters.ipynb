# Arguments Type:
1. Positional argument
2. Keyword argument
3. Default argument
4. Variable length argument

Positional argument

```python
def add_num(x, y):
    print(x)                               #Positional arugument

add_num (10, 30)
```

Keyword argument

```python
def personal(job, income):
    print('job = ', job)
    print('income = ', income)

personal(income = 200000, job = 'Pharmacist')           #used key word
```

Default argument

```python
def personal(job, income= 10000):
    print('job = ', job)
    print('income = ', income)

personal(job = 'Pharmacist', income = 200000)         #overwrite values
```

Variable length argument

```python
def add_num(*x):
    print(x)
    result = sum(x)
    return result
add_num(10, 30, 40, 100, 1000, 4000)
```

```python
def add_num(x, *y):
    print(x)
    print(y)
add_num(10, 30, 40, 100, 1000, 4000)
```
