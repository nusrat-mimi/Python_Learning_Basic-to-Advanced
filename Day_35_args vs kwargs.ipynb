# args and *kwargs
keyworded variable-length argument (**kwargs)

Keyword argumets

```python
def personal(job, income):
    print('job = ', job)
    print('income = ', income)

personal(income = 200000, job = 'Pharmacist') 
```

args and *kwargs  
Name: 'Nusrat Mimi'  
Var: 'Freie'  
City: Berlin

variable length argument

```python
def personal_data(name, *arg):
    print(name)
    print(arg)
personal_data('Nusrat Mimi', 'Freie', 'Berlin')
```

```python
def personal_data(name, *arg):
    print(name)
    print(arg)
    print(type(arg))
personal_data('Nusrat Mimi', 'Berlin', 'Freie')          #using * take tupple 
```

Key worded variable length argument

```python
def personal_data(name, **arg):
    print(name)
    print(arg)
    print(type(arg))
personal_data('Nusrat Mimi', City ='Berlin',Var = 'Freie')         #using ** will take dictionary        
```
