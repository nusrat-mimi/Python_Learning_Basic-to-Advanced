# Example_1: Reading text from a file

```python
with open(r'C:\Users\49176\Downloads\Python Learning_Study mart\Day_41\py1.txt', 'r') as file:
    #read each line one at a time in the file
    for line in file:
        #process each line
        print(line.strip())      #The strip() method removes any leading/trailing whitespce and new line characters
```
